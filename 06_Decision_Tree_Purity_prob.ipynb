#### 1. Outlook

````python
Overcast >> 4Y & 0N
Rainy >> 3Y & 2N
Sunny >> 2Y & 3N

Total Samples = 9Y & 5N >> 14

H(s) = 0.94

H(out==Overcast) = 0
H(out==Rainy) = 0.971
H(out==Sunny) = 0.971

Information = 0.693

IG_Outlook = H(S) - Information
           = 0.94 - 0.693 = 0.247



Excellent! This is the classic **Play Tennis** dataset used to explain **Information Gain** in Decision Trees.

Let's understand **where every number comes from** instead of just memorizing it.

---

## Step 1: Parent Dataset

Suppose we have 14 samples.

| Class | Count |
| ----- | ----: |
| Yes   |     9 |
| No    |     5 |
| Total |    14 |

So,

[
P(Yes)=\frac{9}{14}
]

[
P(No)=\frac{5}{14}
]

---

## Step 2: Calculate Parent Entropy

Formula:

[
H(S)= -P(Y)\log_2(P(Y))-P(N)\log_2(P(N))
]

Substitute:

[
H(S)= -\frac{9}{14}\log_2\left(\frac{9}{14}\right)
-\frac{5}{14}\log_2\left(\frac{5}{14}\right)
]

Result:

[
H(S)=0.940
]

This is the impurity before any split.

---

## Step 3: Split by Outlook

The Outlook feature has three values.

### Overcast

| Yes | No |
| --: | -: |
|   4 |  0 |

Since all samples are **Yes**,

Entropy:

[
H(\text{Overcast})=0
]

Why?

Because the node is perfectly pure.

---

### Rainy

| Yes | No |
| --: | -: |
|   3 |  2 |

Entropy:

[
H(\text{Rainy})=0.971
]

Mixed classes → higher impurity.

---

### Sunny

| Yes | No |
| --: | -: |
|   2 |  3 |

Entropy:

[
H(\text{Sunny})=0.971
]

Also mixed.

---

## Step 4: Calculate Weighted Entropy

Each child node contributes according to **its size**.

### Overcast

Samples:

```text
4 out of 14
```

Contribution:

[
\frac{4}{14}\times0
=0
]

---

### Rainy

Samples:

```text
5 out of 14
```

Contribution:

[
\frac{5}{14}\times0.971
=0.347
]

---

### Sunny

Samples:

```text
5 out of 14
```

Contribution:

[
\frac{5}{14}\times0.971
=0.347
]

---

### Total Weighted Entropy

[
0+0.347+0.347
=============

0.694
]

Rounded:

[
0.693
]

This is what your notes call **Information**, but a better name is:

> **Weighted Entropy after the split**

---

## Step 5: Information Gain

Formula:

[
IG
==

## \text{Parent Entropy}

\text{Weighted Child Entropy}
]

Substitute:

[
IG
==

## 0.940

# 0.693

0.247
]

---

## What does 0.247 mean?

It means:

> **By splitting on the Outlook feature, we reduced the impurity by 0.247.**

The higher the Information Gain, the better the split.

If another feature has:

```text
IG(Humidity) = 0.15
IG(Wind) = 0.05
IG(Outlook) = 0.247
```

Then we choose:

✅ **Outlook** as the root node because it gives the **highest Information Gain**.

---

# Easy Memory Trick

```text
Before Split
-------------
Entropy = 0.94
(More mixed)

        │
 Split by Outlook
        ▼

Overcast → Pure
Rainy     → Mixed
Sunny     → Mixed

Weighted Entropy = 0.693

Information Gain

0.94 - 0.693

= 0.247
```

---

## One important correction

Your notes say:

> **Information = 0.693**

A more accurate term is:

> **Weighted Entropy = 0.693**

because **Information Gain** is actually:

[
\boxed{IG = 0.247}
]

So don't confuse **Weighted Entropy (0.693)** with **Information Gain (0.247)**.

---

### Interview tip

Remember this sequence:

1. Calculate **Parent Entropy**.
2. Split on a feature.
3. Calculate **Entropy of each child node**.
4. Compute the **Weighted Average Entropy**.
5. **Information Gain = Parent Entropy − Weighted Child Entropy**.
6. Choose the feature with the **highest Information Gain** as the decision node.
````

````python
These notes are about **Decision Tree Pruning**. Here's a simple explanation.

---

# What is Pruning?

**Pruning** is the process of **removing unnecessary branches (nodes)** from a Decision Tree.

The goal is to:

* Reduce the complexity of the tree.
* Prevent **overfitting**.
* Improve performance on new (unseen) data.

Think of it like **cutting unnecessary branches from a real tree** to make it healthier.

---

## Why do we need Pruning?

Suppose a Decision Tree grows like this:

```text
                Root
               /    \
             A        B
            / \      / \
           C   D    E   F
              / \
             G   H
```

Some branches may be created because of **noise** or just a few training samples.

These extra branches memorize the training data instead of learning general patterns.

This leads to **overfitting**.

Pruning removes such branches.

---

# Types of Pruning

There are **two types**:

## 1. Pre-Pruning (Early Stopping)

The tree is stopped **before** it becomes too large.

Example:

```text
          Root
         /    \
       A        B
```

The algorithm decides:

> "This tree is already good enough. Stop growing."

It never creates unnecessary nodes.

### Common parameters

```python
DecisionTreeClassifier(
    max_depth=3,
    min_samples_split=10,
    min_samples_leaf=5
)
```

Most commonly used:

```python
max_depth
```

### Advantages

* Faster training.
* Smaller tree.
* Less memory.

### Disadvantage

* It may stop **too early**, causing **underfitting**.

---

## 2. Post-Pruning

First, build the **full tree**.

```text
                Root
               /    \
             A        B
            / \      / \
           C   D    E   F
              / \
             G   H
```

Then remove branches that don't improve the model.

After pruning:

```text
          Root
         /    \
       A        B
```

This usually produces a better balance between complexity and accuracy.

---

### In Scikit-learn

Use:

```python
DecisionTreeClassifier(
    ccp_alpha=0.01
)
```

`ccp_alpha` stands for:

> **Cost Complexity Pruning Alpha**

---

### How does `ccp_alpha` work?

* **ccp_alpha = 0**

  * No pruning.
  * Full tree.

* **Larger `ccp_alpha`**

  * More branches are removed.
  * Simpler tree.

---

# Comparison

| Pre-Pruning                                               | Post-Pruning                                                  |
| --------------------------------------------------------- | ------------------------------------------------------------- |
| Stops tree growth early                                   | Builds the full tree first, then removes unnecessary branches |
| Faster                                                    | Slightly slower                                               |
| Uses `max_depth`, `min_samples_split`, `min_samples_leaf` | Uses `ccp_alpha`                                              |
| May underfit if stopped too soon                          | Often gives better generalization                             |

---

# Real-life Analogy

Imagine you're writing notes for an exam.

### Without Pruning

You write **every single detail** from the textbook.

📚 500 pages.

You memorize everything, including unimportant details.

This is like **overfitting**.

---

### Pre-Pruning

Your teacher says:

> "Stop after Chapter 5."

You stop early.

Maybe you miss some useful topics.

This can cause **underfitting**.

---

### Post-Pruning

You first write all 500 pages.

Then you remove unnecessary or repeated information and keep only what matters.

Now you have a concise, effective set of notes.

This is usually the better approach.

---

# Interview Answer (30 seconds)

> **Pruning** is the process of removing unnecessary branches from a Decision Tree to reduce complexity and prevent overfitting. There are two types: **Pre-Pruning**, where tree growth is stopped early using parameters like `max_depth`, `min_samples_split`, or `min_samples_leaf`, and **Post-Pruning**, where the tree is first grown completely and then simplified using **Cost Complexity Pruning (`ccp_alpha`)**. Pre-pruning is faster but may underfit, while post-pruning generally produces a better-balanced model that generalizes well to unseen data.
````

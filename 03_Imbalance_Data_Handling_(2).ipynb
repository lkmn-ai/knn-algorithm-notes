````python
Your notes are on the right track. Here's a cleaner version with the correct terminology and examples.

---

# Handling Imbalanced Data

An **imbalanced dataset** is one where one class has significantly more samples than the other.

### Example

| Class                | Samples |
| -------------------- | ------: |
| Fraud (Minority)     |     150 |
| Not Fraud (Majority) |     850 |

The model may become biased toward the majority class.

---

# Techniques to Handle Imbalanced Data

There are two main approaches:

1. **Undersampling**
2. **Oversampling**

---

# 1. Undersampling

Undersampling means **reducing the number of samples in the majority class** to make the dataset more balanced.

Example:

Original dataset:

```text
Fraud (Minority)      = 150
Not Fraud (Majority)  = 850
```

After undersampling:

```text
Fraud      = 150
Not Fraud  = 150
```

or

```text
Fraud      = 150
Not Fraud  = 250
```

### Advantages

* Faster training.
* Balanced dataset.

### Disadvantages

* ❌ Important information may be lost because many majority-class samples are discarded.
* ❌ Not recommended when the dataset is already small.

**Rule:** Avoid undersampling if you have limited data.

---

# 2. Oversampling

Oversampling means **increasing the number of samples in the minority class**.

There are two common ways:

### (a) Random Oversampling

Simply duplicate existing minority samples.

Example:

Original:

```text
Fraud      = 150
Not Fraud  = 850
```

After oversampling:

```text
Fraud      = 850
Not Fraud  = 850
```

The extra fraud samples are just repeated copies.

### Advantage

* No information is lost.

### Disadvantage

* The model may **overfit** because it keeps seeing the same minority examples.

---

### (b) Synthetic Oversampling (SMOTE)

Instead of copying existing samples, SMOTE **creates new artificial minority samples** by interpolating between existing minority samples.

Example:

Original minority samples:

```text
(10, 20)
(12, 22)
```

SMOTE might generate:

```text
(11, 21)
```

This new point didn't exist before but is realistic because it lies between similar minority samples.

### Advantages

* No loss of data.
* Less overfitting than random oversampling.
* Often gives better performance.

---

# Summary Table

| Technique           | What it does                       | Advantage             | Disadvantage                                       |
| ------------------- | ---------------------------------- | --------------------- | -------------------------------------------------- |
| Undersampling       | Removes majority-class samples     | Faster training       | May lose important information                     |
| Random Oversampling | Duplicates minority-class samples  | No data loss          | Can overfit                                        |
| SMOTE               | Creates synthetic minority samples | Better generalization | Can create noisy samples if minority data is noisy |

---

# Interview Answer (30 seconds)

> **Imbalanced data** occurs when one class has significantly more samples than another. Two common ways to handle it are **undersampling** and **oversampling**. **Undersampling** reduces the majority class to balance the dataset, but it may discard useful information. **Oversampling** increases the minority class, either by **duplicating existing samples** (Random Oversampling) or by **creating synthetic samples** using techniques like **SMOTE**. In practice, oversampling with SMOTE is often preferred because it preserves the majority-class data while reducing the risk of overfitting compared to simple duplication.
````

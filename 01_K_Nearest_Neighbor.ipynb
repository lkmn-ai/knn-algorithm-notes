## K-Nearest Neighbor

````python
It is a distance based algorithm
Non-Probabilistic algorithm

Feature Scaling is required.
It is used for Classification as well as Regression.
But mostly used for Classification.

It works on Feature Similarity Measure.
It is a Lazy Learner, Because it doesnt learn from training data,
Instead it stores all the available data and classifies at the time of prediction.

Accuracy will increase when training data is big
But complexity problem could be there.

It is a versatile algorithm, also used for:
    1. KNN Imputer
    2. Imbalance Data Handling (Resampling)
````

### KNN Algorithm

````python
1. Load the Data
2. Select the k-value (no of k neighbors)
3. Calculate the distance of Testing(New) datapoint with all datapoints
4. Sort distances in ascending order
5. Select first k close distance neighbors(datapoints) from sorted values
6. Classification:
    Voting: Majority class will be predicted class
7. Regression:
    Mean of first k neighbors
````

### Industry Application of KNN

````python
1. Recommendation System(Amazon, YouTube)
2. Document Classification
````

### Advantages

````python
1. The algorithm is easy and simple to understand and implement
2. This algorithm is versatile. It can be used for Classification and Regression:
    Versatile:
        KNN Imputer
        SMOTE for Oversampling
3. Work with any number of classes, not just binary classifiers
4. Efficient for small datasets
5. Non-Parametric (No assumption on data)
6. Works well on Non-Linear Data
````

### Disadvantages

````python
1. Data Scaling(Feature Scaling) is required:
    1. Normalization
    2. Standardization
2. The compuatational cost is high because of calculating the distances between datapoints for all training samples
3. Always needs to determine value of k, which can be complex sometimes
4. Doesnt work as expected when working with big number of features (High Dimensional Data) (It takes more time to process):
    a. Select Best Feature:
        1. Feature Selection
        2. Feature Extraction : Principal Component Analysis
5. Noisy data can result in over-fitting or under-fitting:
    Over-fitting >> if k value is low
    Under-fitting >> if k value is high
6. Sensitive to outliers
````

````python
The data scientist or ML engineer chooses the value of K. It is a hyperparameter that is not learned from the data.
We usually try multiple values of K and use a validation set or cross-validation to choose the one that gives the best model performance.
A small K can overfit and be sensitive to noise, while a very large K can underfit and miss local patterns.
````

````python
Let's understand KNN with a simple real-world example.

### Example: Predicting whether a fruit is an Apple or an Orange

Suppose our dataset has two features:

* **Weight (grams)**
* **Color Score** (0 = Green, 10 = Deep Red)

| Fruit  | Weight (g) | Color Score | Class  |
| ------ | ---------: | ----------: | ------ |
| Apple  |        150 |           8 | Apple  |
| Apple  |        170 |           9 | Apple  |
| Apple  |        140 |           7 | Apple  |
| Orange |        130 |           4 | Orange |
| Orange |        120 |           3 | Orange |
| Orange |        135 |           5 | Orange |

Now a new fruit arrives:

* **Weight = 145 g**
* **Color Score = 7**

We want to classify it.

### Step 1: Choose K

Suppose **K = 3**.

This means we look at the **3 nearest neighbors**.

### Step 2: Calculate Distances

Using Euclidean distance:

[
\text{Distance}=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
]

Distances:

| Training Fruit | Distance | Class  |
| -------------- | -------: | ------ |
| Apple (150,8)  |     5.10 | Apple  |
| Apple (170,9)  |    25.08 | Apple  |
| Apple (140,7)  |     5.00 | Apple  |
| Orange (130,4) |    15.30 | Orange |
| Orange (120,3) |    25.32 | Orange |
| Orange (135,5) |    10.20 | Orange |

### Step 3: Select the 3 Nearest Neighbors

| Neighbor       | Class  |
| -------------- | ------ |
| Apple (140,7)  | Apple  |
| Apple (150,8)  | Apple  |
| Orange (135,5) | Orange |

### Step 4: Majority Voting

* Apple → **2 votes**
* Orange → **1 vote**

✅ **Prediction = Apple**

---

## Visual Representation

```
Color Score ↑

9 |                  A
8 |             A
7 |         A      X (New Fruit)
6 |
5 |      O
4 |    O
3 |  O
  +----------------------------→ Weight
    120 130 140 150 160 170
```

* **A = Apple**
* **O = Orange**
* **X = New fruit**

The new fruit (**X**) is closer to the Apple points than the Orange points, so KNN predicts **Apple**.

---

## Real-life Analogy

Imagine you join a new college and want to guess whether a student is from the **CSE** or **ECE** department.

You look at the **5 students sitting closest** to them:

* CSE
* CSE
* ECE
* CSE
* CSE

Since **4 out of 5** nearest students are from **CSE**, you predict the new student is also from **CSE**.

This is exactly how **KNN Classification** works—it assumes that **similar (nearby) data points usually belong to the same class**.

---

### Why is it called **Lazy Learning**?

During training, KNN **does not learn any equations or build a model**. It simply stores the training data:

```
Training Phase:
Store all points ✔

Prediction Phase:
        ↓
Receive new point
        ↓
Calculate distance to every training point
        ↓
Pick K nearest neighbors
        ↓
Majority vote
        ↓
Predict class
```

That's why KNN has **almost no training time**, but **prediction becomes slower** as the training dataset grows.
````

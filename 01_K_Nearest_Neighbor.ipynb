## K-Nearest Neighbor

```python
It is a distance based algorithm
Non-Probabilistic algorithm

Feature Scaling is required.
It is used for Classification as well as Regression.
But mostly used for Classification.

It works on Feature Similarity Measure.
It is a Lazy Learner, Because it doesnt learn from training data,
Instead it stores all the available data and classifies at the time of prediction.

Accuracy will increase when training data is big
But complexity problem could be there.

It is a versatile algorithm, also used for:
    1. KNN Imputer
    2. Imbalance Data Handling (Resampling)
```

### KNN Algorithm

```python
1. Load the Data
2. Select the k-value (no of k neighbors)
3. Calculate the distance of Testing(New) datapoint with all datapoints
4. Sort distances in ascending order
5. Select first k close distance neighbors(datapoints) from sorted values
6. Classification:
    Voting: Majority class will be predicted class
7. Regression:
    Mean of first k neighbors
```

### Industry Application of KNN

```python
1. Recommendation System(Amazon, YouTube)
2. Document Classification
```

### Advantages

```python
1. The algorithm is easy and simple to understand and implement
2. This algorithm is versatile. It can be used for Classification and Regression:
    Versatile:
        KNN Imputer
        SMOTE for Oversampling
3. Work with any number of classes, not just binary classifiers
4. Efficient for small datasets
5. Non-Parametric (No assumption on data)
6. Works well on Non-Linear Data
```

### Disadvantages

```python
1. Data Scaling(Feature Scaling) is required:
    1. Normalization
    2. Standardization
2. The compuatational cost is high because of calculating the distances between datapoints for all training samples
3. Always needs to determine value of k, which can be complex sometimes
4. Doesnt work as expected when working with big number of features (High Dimensional Data) (It takes more time to process):
    a. Select Best Feature:
        1. Feature Selection
        2. Feature Extraction : Principal Component Analysis
5. Noisy data can result in over-fitting or under-fitting:
    Over-fitting >> if k value is low
    Under-fitting >> if k value is high
6. Sensitive to outliers
```

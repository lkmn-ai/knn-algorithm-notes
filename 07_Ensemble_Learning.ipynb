## Random Forest


```python
Random Forest:
    Combining Multiple Decision Trees
    It is used for Classification as well as Regression
    Create boostrap datasets by using row sampling(bootstrap means random subsets of data which we take those data to train it)
    Classification:
        Default >> Hard (Majority)
    Regression:
        Mean of all Decision Trees

    RF Model converts low bias and high variance models to a single low bias and low variance model

Out of Bag Samples(Leftover):
    Samples which are not used by bootstrapped datasets



Real-life Analogy for Bagging

Imagine asking one doctor.He may make a wrong diagnosis.

Now ask 10 doctors.If 9 doctors say the same thing,

their combined opinion is much more reliable.Each doctors individual mistake is averaged out.Thats exactly how Bagging reduces variance

Q. Why is Random Forest called a Bagging algorithm?
Because it creates multiple Decision Trees using bootstrap samples (random subsets of the training data)
and combines their predictions using majority voting (classification) or averaging (regression).

See bagging is taking all the models into consideration parallely and predicting the outcome so variance, if we change the dataset in a small quantity only model will
behave differently, so by taking many opinions we can reduce the variance right


| Bagging                                  | Boosting                                 |
| ---------------------------------------- | ---------------------------------------- |
| Many models are trained independently.   | Models are trained sequentially.         |
| Each model gets a random subset of data. | Each model focuses on previous mistakes. |
| Reduces **Variance** (less overfitting). | Reduces **Bias** (less underfitting).    |
| Example: Random Forest                   | Example: AdaBoost, XGBoost               |
```

#### Advantages   

```python
1. It is used for Classification as well as Regression
2. It helps to reduce overfitting:
    by reducing variance
    improve the model performance
3. RF will perform better on contineous data as well as categorical data
4. No feature scaling is required
5. Less impacted by outliers
6. Non-Parametric algo (No assumption on distribution of data):
    Peforms well on non-linear data
7. RF Models is very stable as compared to DT:
    Not impacted by adding or deleting datapoints in dataset
```

#### Disadvantages

```python
1. Longer training time (Requires more computational power)
2. Complexity:
    100 different decision trees
```

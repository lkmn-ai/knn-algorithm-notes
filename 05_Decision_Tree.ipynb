````python
Your notes are very good. I'll organize them and explain the concepts in a way that's easy to understand and remember.

---

# Decision Tree

A **Decision Tree** is a **Supervised Machine Learning algorithm** used for:

* ✅ Classification
* ✅ Regression

It predicts the output by asking a sequence of **if-else questions**.

---

## CART

**CART** stands for:

> **Classification And Regression Trees**

* Classification → Predicts categories (Yes/No, Spam/Not Spam)
* Regression → Predicts continuous values (House Price, Salary)

---

## Example

Suppose a bank wants to decide whether to approve a loan.

```text
            Income > 50K?
               /      \
             No        Yes
            /            \
      Reject      Credit Score > 700?
                      /         \
                    No          Yes
                   /              \
              Reject           Approve
```

This is a Decision Tree.

Each decision is based on a condition.

---

# Components of Decision Tree

### 1. Root Node

The first node.

Example:

```text
Income > 50K?
```

It is the first decision.

---

### 2. Branch

The arrows connecting nodes.

```text
Yes
No
```

These represent the possible outcomes of a decision.

---

### 3. Internal (Decision) Node

Any node that asks another question.

Example:

```text
Credit Score > 700?
```

---

### 4. Leaf Node (Terminal Node)

The final prediction.

```text
Approve Loan

Reject Loan
```

No more questions.

---

# Purity vs Impurity

Imagine we want to classify fruits.

### Pure Node

```text
Apple
Apple
Apple
Apple
```

Only one class.

Impurity = **0**

Perfect node.

---

### Impure Node

```text
Apple
Apple
Orange
Orange
```

Mixed classes.

High impurity.

---

Decision Trees try to make every leaf node as **pure as possible**.

---

# How does the tree decide which question to ask?

Suppose we have two possible questions.

Question 1

```text
Age > 30?
```

Question 2

```text
Salary > 50000?
```

Which one should become the root node?

The algorithm checks **which question reduces impurity the most**.

The one that creates the **purest child nodes** is selected.

---

# Measures of Impurity

There are two common ways to measure impurity.

## 1. Gini Impurity

Used in **CART**.

Formula:

[
\text{Gini} = 1 - \sum p_i^2
]

For binary classification:

[
\text{Gini}=1-p^2-q^2
]

where:

* p = probability of Yes
* q = probability of No

### Interpretation

* Gini = 0 → Perfectly pure.
* Higher Gini → More mixed classes.

---

## 2. Entropy

Used in **ID3**.

Formula:

[
H(S) = -P(Y)\log_2P(Y)-P(N)\log_2P(N)
]

Entropy measures **uncertainty**.

* Entropy = 0 → Pure.
* Higher entropy → More disorder.

---

### Example

Suppose a node contains:

```text
Yes
Yes
Yes
Yes
```

Entropy:

```text
0
```

Perfectly pure.

---

Suppose:

```text
Yes
Yes
No
No
```

Entropy becomes:

```text
1
```

Maximum uncertainty for two equally likely classes.

---

# Information Gain (IG)

Information Gain tells us:

> **How much uncertainty (entropy) decreases after splitting the data.**

Formula:

[
IG = \text{Entropy(Parent)} - \text{Weighted Entropy(Children)}
]

where

[
\text{Weighted Entropy}=
\sum
\frac{\text{Child Samples}}
{\text{Parent Samples}}
\times
\text{Entropy(Child)}
]

---

## Example

Before splitting

```text
10 Apples
10 Oranges
```

Entropy = High.

---

Split using:

```text
Color = Red?
```

Left:

```text
10 Apples
```

Entropy = 0

Right:

```text
10 Oranges
```

Entropy = 0

Information Gain is very high because the split created two perfectly pure nodes.

---

# Algorithms

### CART

* Uses **Gini Impurity**
* Supports Classification and Regression
* Binary splits (each node has two children)

---

### ID3

* Uses **Entropy**
* Classification only
* Can create multiple branches from a node

---

# Interview Table

| Algorithm | Uses       | Split Type      |
| --------- | ---------- | --------------- |
| CART      | Gini Index | Binary split    |
| ID3       | Entropy    | Multi-way split |

---

# Easy Way to Remember

Imagine a classroom.

Before grouping:

```text
Math
Science
Math
Science
Math
Science
```

Very mixed (high impurity).

Now group by subject.

Room 1:

```text
Math
Math
Math
```

Room 2:

```text
Science
Science
Science
```

Now both rooms are **pure**.

A Decision Tree keeps asking questions to split the data until each group is as pure as possible.

---

## One correction to your notes

You wrote:

> **Information Gain: IG = Entropy - (weighted avg * Entropy of each feature)**

A clearer and more accurate version is:

[
\boxed{\text{Information Gain} = \text{Entropy of Parent} - \text{Weighted Average Entropy of Child Nodes}}
]

The "weighted average" is based on **how many samples go into each child node**, not on the features themselves.

---

## Interview Answer (30 seconds)

> A **Decision Tree** is a supervised learning algorithm used for both **classification** and **regression**. It predicts by recursively splitting the data based on feature values. The algorithm selects the best split by minimizing impurity. For classification, impurity is commonly measured using **Gini Impurity** (used by CART) or **Entropy** (used by ID3). The best split is chosen using **Information Gain**, which measures how much the entropy decreases after the split. The objective is to create leaf nodes that are as **pure** as possible, meaning they contain samples from a single class.
````

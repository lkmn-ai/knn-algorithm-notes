## Feature Scaling

````python
Feature Scaling:
    It is a statistical method/technique used to normalize / standardize the independent features of data in a range.

Scaling Methods:
    1. Normalization
    2. Standardization

Types of Feature Scaling:
    1. MinMax Scaler (Normalization)
    2. Standard Scaler (Standardization)
    3. Robust Scaler >> Effective on outliers
    4. Max Absolute
    5. Unit Vector Scaler
    6. Power Transform
    7. Quantile Transform
    8. Mean Normalization

Scaling is not required on Target Columns and One Hot Encoded Columns or Binary Columns.
````

````python
Feature Scaling is a preprocessing technique used to bring independent features to a common scale so that features with larger values do not dominate those with smaller values.
The two most common methods are Normalization (Min-Max Scaling) and Standardization (Z-score Scaling). Scaling is important for distance-based and
optimization-based algorithms like KNN, K-Means, SVM, PCA, and Neural Networks, but it is generally not required for tree-based algorithms such as Decision Trees and Random Forests.
Typically, only the feature columns (X) are scaled—not the target column or one-hot encoded/binary columns (except in some regression scenarios where scaling the target can be beneficial).
````

````python
This is one of the most commonly asked interview questions. Don't memorize it—understand **why**.

---

# When to use Normalization vs Standardization?

| Normalization (Min-Max Scaling)                | Standardization (StandardScaler)                     |
| ---------------------------------------------- | ---------------------------------------------------- |
| Scales data between **0 and 1**                | Mean = **0**, Std = **1**                            |
| Preserves the shape of the data                | Centers the data around the mean                     |
| Sensitive to outliers                          | Sensitive to outliers                                |
| Best when data has a known minimum and maximum | Best when data is approximately normally distributed |

---

## Use **Normalization** when

### 1. The algorithm uses distances

Example:

* KNN
* K-Means

### Why?

Suppose you have:

| Age | Salary |
| --- | ------ |
| 25  | 25,000 |
| 30  | 90,000 |

Without scaling:

```
Age difference = 5

Salary difference = 65000
```

Salary dominates the distance.

Normalization converts them to:

| Age | Salary |
| --- | ------ |
| 0.2 | 0.1    |
| 0.5 | 0.9    |

Now both features contribute equally.

---

### Algorithms that usually use Normalization

* ✅ KNN
* ✅ K-Means
* ✅ Neural Networks (commonly)
* ✅ Image Processing (pixel values 0–255 → 0–1)

---

## Use **Standardization** when

### Data is approximately normally distributed.

Example:

Heights of students

```
160
165
170
175
180
```

Here we care about:

> How far is a student's height from the average?

StandardScaler answers exactly that.

---

### Algorithms that commonly use Standardization

* ✅ Logistic Regression
* ✅ Linear Regression (often beneficial)
* ✅ SVM
* ✅ PCA
* ✅ LDA

---

# Example

Suppose salaries are:

```
20000
30000
40000
50000
60000
```

### Normalization

```
0
0.25
0.5
0.75
1
```

Everything lies between **0 and 1**.

---

### Standardization

Suppose:

Mean = 40000

Std = 10000

Then

```
20000 → -2
30000 → -1
40000 → 0
50000 → 1
60000 → 2
```

Now the values tell us **how many standard deviations** they are from the mean.

---

# What if the data has outliers?

Suppose:

```
20
25
30
35
1000
```

### Normalization

Since the maximum is **1000**, almost every other value becomes very close to **0**.

Example:

```
20  → 0
25  → 0.005
30  → 0.01
35  → 0.015
1000→ 1
```

The normal values get compressed.

### Standardization

StandardScaler is also affected because the **mean** and **standard deviation** increase due to the outlier.

So if you have **many outliers**, **don't use either MinMaxScaler or StandardScaler**. Instead, use **RobustScaler**, which is based on the **median** and **IQR**.

---

# Easy Interview Rule

```
Is the algorithm distance-based?
        │
       Yes
        │
Use Scaling
        │
        ├── Data approximately normal?
        │        │
        │       Yes → StandardScaler
        │
        └── Need values between 0 and 1?
                 │
                Yes → MinMaxScaler
```

---

## One thing many beginners ask

> **If KNN is distance-based, why do many tutorials use MinMaxScaler instead of StandardScaler?**

Because KNN only requires **features to be on a comparable scale**. Both **MinMaxScaler** and **StandardScaler** work well. If the data is approximately normal, **StandardScaler** is often preferred; if you want features bounded between **0 and 1** (or you're working with image pixels or neural networks), **MinMaxScaler** is a common choice.

---

### Interview answer (30 seconds)

> **Normalization** scales features to a fixed range, usually **0 to 1**, and is commonly used for algorithms like **KNN, K-Means, and Neural Networks**, especially when bounded feature values are helpful. **Standardization** transforms features to have a **mean of 0** and a **standard deviation of 1**, making it a good choice when the data is approximately normally distributed and for algorithms such as **Logistic Regression, SVM, PCA, and Linear Regression**.
                                                                                                                                                                                                                                                          If the dataset contains significant outliers, **RobustScaler** is generally a better option than either MinMaxScaler or StandardScaler.
````
